```python
import pandas as pd
import os

from dotenv import load_dotenv, find_dotenv
from sqlalchemy import create_engine, URL

# Load environment variables
load_dotenv(find_dotenv(usecwd=True))

# Get MySQL password from .env
password = os.getenv("MYSQL_PASSWORD")

if not password:
    raise ValueError("MYSQL_PASSWORD not found in .env file")

# Create MySQL connection
connection_url = URL.create(
    drivername="mysql+pymysql",
    username="root",
    password=password,
    host="localhost",
    port=3306,
    database="salesdb"
)

engine = create_engine(connection_url)

# Load cleaned dataset
df = pd.read_csv('../data/processed/Sales_Cleaned_dataset.csv')

# Upload data to MySQL
df.to_sql(
    name='sales_data',
    con=engine,
    if_exists='replace',
    index=False
)

print("Data successfully uploaded to MySQL!")
```

```python
from sqlalchemy import text

with engine.connect() as connection:

    # Check total records
    result = connection.execute(
        text("SELECT COUNT(*) FROM sales_data")
    )

    print("Total records:", result.scalar())

    # Display first 5 records
    result = connection.execute(
        text("SELECT * FROM sales_data LIMIT 5")
    )

    for row in result:
        print(row)
```
